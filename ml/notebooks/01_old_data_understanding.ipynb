```python
df_old["timestamp"] = pd.to_datetime(
    df_old["timestamp"],
    errors="coerce"
)
```

```python
print(df_old["timestamp"].min())
print(df_old["timestamp"].max())
```

```python
df_old = df_old.sort_values("timestamp").reset_index(drop=True)
```

```python
sensor_columns = [
    col for col in df_old.columns
    if col.startswith("sensor_")
]

sensor_profile = pd.DataFrame({
    "missing_count": df_old[sensor_columns].isnull().sum(),
    "missing_percentage": df_old[sensor_columns].isnull().mean() * 100,
    "unique_values": df_old[sensor_columns].nunique(),
    "mean": df_old[sensor_columns].mean(),
    "std": df_old[sensor_columns].std(),
    "min": df_old[sensor_columns].min(),
    "max": df_old[sensor_columns].max()
})

sensor_profile.sort_values(
    "missing_percentage",
    ascending=False
)
```
